### S3: Model Deployment & Caching

**Question:** Describe the sequence of steps — from saving the model file to displaying the prediction — that makes this Streamlit prediction app work correctly, and explain why wrapping the model load function with st.cache_resource is critical for production performance.

**Answer:**
**Sequence of Steps:**
1.  Load the trained model using `joblib.load('delivery_model.pkl')` inside a custom function.
2.  Decorate this loading function with `@st.cache_resource`.
3.  Create Streamlit input widgets (e.g., `st.number_input`, `st.slider`) to collect order details from the dispatcher.
4.  On a button click (`st.button`), capture the inputs, format them into a 2D array, and pass them to the model's `.predict()` method.
5.  Display the result to the user using `st.success()`.

**Importance of Caching:** Streamlit reruns the entire script from top to bottom on every user interaction. Wrapping the model loader in `st.cache_resource` ensures that the heavy machine learning model is loaded into memory exactly once. Without it, the model would reload from disk every time the dispatcher adjusted a slider, causing severe lag and terrible production performance.

### S4: Activation Functions for Classification

**Question:** Which activation function should the output layer use for this three-class classification task, and why does using sigmoid on the output layer produce incorrect probability distributions when you need mutually exclusive class probabilities across all three tiers?

**Answer:**
The output layer should use the **`softmax`** activation function.

**Why Sigmoid is Incorrect:** The sigmoid function treats every output node completely independently, squashing each score between 0 and 1. It does not constrain the outputs to sum to 100%. This means the model could predict an 85% probability for Standard AND a 75% probability for Priority at the same time. Softmax, on the other hand, normalizes the outputs into a strict probability distribution that always sums to 1, accurately enforcing the mutual exclusivity required for this task.

### S5: Regularization Strategies

**Question:** Compare Dropout regularization and L2 weight regularization as strategies to reduce this gap. For a dense multi-layer neural network, which would you apply first in practice, at what typical rate or strength, and what different mechanism does each use to discourage overfitting?

**Answer:**
*   **Dropout:** Randomly "drops" or disables a percentage of neurons during each training step. Mechanism: It prevents neurons from co-adapting and forces the network to distribute learning across multiple redundant pathways.
*   **L2 Regularization (Weight Decay):** Adds a penalty term to the loss function based on the squared magnitude of the network's weights. Mechanism: It discourages the network from assigning excessively large weights to any single feature, keeping the decision boundaries smooth and generalized.

**In Practice:** **Dropout** is usually applied first because it is highly effective and computationally cheap for Dense networks. A typical starting rate is **0.2 to 0.5**, added immediately after the activation functions of the hidden layers.

### S6: ANN vs CNN Architecture

**Question:** Explain the key architectural difference between an ANN and a CNN that makes CNNs significantly more effective for image classification, and describe what a convolutional filter learns from spatial data that fully connected layers cannot efficiently capture.

**Answer:**
**Architectural Difference:** An ANN requires flattening the 2D image into a 1D vector before processing, which completely destroys the spatial relationships between neighboring pixels. A CNN preserves the 2D grid structure and uses convolutional layers to process spatial chunks directly.

**What Filters Learn:** A convolutional filter slides across the image to learn local spatial patterns like edges, corners, and textures. Because the same filter is applied across the entire image, it achieves "translation invariance"—meaning it can recognize an object regardless of whether it is in the top-left or bottom-right corner. Fully connected layers treat every pixel as an independent feature and struggle massively to capture these spatial hierarchies efficiently.
